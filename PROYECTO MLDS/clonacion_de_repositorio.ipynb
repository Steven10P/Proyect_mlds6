```python
%pip install databricks-sdk --quiet
dbutils.library.restartPython()

from databricks.sdk import WorkspaceClient

# Inicializa el cliente (usa tus credenciales actuales de Databricks)
w = WorkspaceClient()

repo_url = "https://github.com/Steven10P/Proyect_mlds6.git"
# Define dónde guardarlo (usualmente en tu carpeta de usuario)
repo_path = f"/Workspace/Users/{w.current_user.me().user_name}/Proyect_mlds6_git"

print(f"Clonando {repo_url} en {repo_path}...")

try:
    # Intenta crear el Repo
    repo = w.repos.create(
        url=repo_url,
        provider="gitHub",
        path=repo_path
    )
    print(f"✅ Éxito! ID del Repo: {repo.id}")
except Exception as e:
    print(f"⚠️ Nota: Si ya existe o hay error: {e}")
```
